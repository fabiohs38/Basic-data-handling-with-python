```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calcular a receita média para cada categoria em ordem decrescente
category_order = df.groupby('category')['revenue_usd'].mean().sort_values(ascending=False).index

# Criar o gráfico de barras
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='category', y='revenue_usd', order=category_order, ci=None)
plt.title('Receita Média por Categoria de Produto')
plt.xlabel('Categoria')
plt.ylabel('Receita Média (USD)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

```

### 1. Gráfico de Barras com Fonte de Tráfego (`hue`)


```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='category', y='revenue_usd', order=category_order, hue='traffic_source', errorbar=None)
plt.title('Receita Média por Categoria e Fonte de Tráfego')
plt.xlabel('Categoria')
plt.ylabel('Receita Média (USD)')
plt.legend(title='Fonte de Tráfego')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 2. Gráfico de Dispersão: Itens Vendidos vs Receita
Análise da relação entre a quantidade de itens em um pedido (`units_sold`) e a receita (`revenue_usd`).

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='units_sold', y='revenue_usd', s=100)
plt.title('Relação entre Itens Vendidos e Receita')
plt.xlabel('Unidades Vendidas (itens por pedido)')
plt.ylabel('Receita (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### 3. Gráfico de Dispersão com Cores por Categoria
Adicionando a categoria do produto como cor utilizando a paleta `'deep'`.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='units_sold', y='revenue_usd', hue='category', palette='deep', s=100)
plt.title('Itens Vendidos vs Receita por Categoria')
plt.xlabel('Unidades Vendidas')
plt.ylabel('Receita (USD)')
plt.legend(title='Categoria')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Análise e Respostas

#### 1. Qual categoria tem as maiores vendas médias? Qual vende menos?
* **Maiores vendas médias:** A categoria de **Electronics** (Eletrônicos) lidera de forma destacada, apresentando a maior receita média por pedido.
* **Menores vendas médias:** A categoria de **Books** (Livros) é a que apresenta o menor desempenho em receita média.

#### 2. Popularidade das fontes de tráfego (ao adicionar o `hue`):
* **Electronics** é altamente impulsionado por tráfego **Direct** (Direto) e **Organic Search** (Busca Orgânica).
* **Furniture** (Móveis) destaca-se majoritariamente através de **Google Ads**.
* **Clothing** (Roupas) e **Books** (Livros) demonstram um apelo equilibrado em **Social Media** (Redes Sociais) e **Organic Search**.
* **Conclusão:** Tráfego direto e buscas orgânicas trazem receitas médias muito robustas para produtos de alto valor unitário (eletrônicos), enquanto campanhas pagas (Google Ads) são o principal motor para categorias específicas como móveis.

#### 3. Natureza da relação entre o número de itens e a receita:
* Há uma **relação inversa/negativa**: pedidos com menos itens (ex: de 5 a 15) tendem a produzir receitas significativamente maiores em comparação com pedidos que possuem muitos itens (ex: de 18 a 30).
* Isso sugere que o valor unitário dos produtos varia consideravelmente, onde itens de alto valor costumam ser comprados individualmente ou em menor quantidade.

#### 4. Diferenças entre as categorias no gráfico de dispersão:
* **Sim, as diferenças são nítidas:**
  * **Electronics (Eletrônicos):** Concentra-se no quadrante superior esquerdo (poucos itens vendidos, porém receitas muito altas). São produtos com elevado valor agregado.
  * **Clothing (Roupas) e Books (Livros):** Concentram-se no quadrante inferior direito (muitos itens por pedido, mas receitas baixas/moderadas). São produtos com baixo valor unitário.
  * **Furniture (Móveis):** Apresenta uma distribuição intermediária (baixa quantidade de itens com receita média/alta).
